```python
import pandas as pd 
import matplotlib.pyplot as plt 
import numpy as np
```

```python
import os
os.makedirs("Sales_charts", exist_ok=True)
```

CREATE OUR SALES DATASET

```python
data = { 
        "Product": ["laptop", "phone", "Headphones", "keyboard", "mouse"],
        "Price": [500000, 300000, 25000, 15000, 8000],
        "Quantity_sold": [5, 10, 20, 15, 30]
}        
df = pd.DataFrame(data)
print(df)
```

CALCULATE THE TOTAL SALE

```python
df["Total_sales"] = df["Price"] * df["Quantity_sold"]
print(df)
```

CALCULATE THE TOTAL SALE REVENUE

```python
Total_sales = df["Total_sales"].sum()
print("Total sales for all products: ₦", Total_sales)
```

FIND THE BEST SELLING PRODUCT

```python
best_selling_product = df.loc[df["Quantity_sold"].idxmax(), "Product"]

print("Best selling product:", best_selling_product)
print("Quantity sold of best selling product:", df.loc[df["Quantity_sold"].idxmax(), "Quantity_sold"])
```

FIND THE PRODUCT THAT EARNED THE MOST MONEY

```python
Highest_sales_product = df.loc[df["Total_sales"].idxmax(), "Product"]

print("Product with highest sales:", Highest_sales_product)
print("Total sales of highest selling product: ₦", df.loc[df["Total_sales"].idxmax(), "Total_sales"])
```

CALCULATE THE REVENUE AVERAGE 

```python
average_sales = df["Total_sales"].mean()
print("Average sales per product: ₦", average_sales)
```

 A BAR CHART OF PRODUCT SALES

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Product"], df["Total_sales"], color='BLACK')
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.title("Total Sales per Product")

plt.savefig("Sales_charts/Total_sale_per_product.png", dpi=300, bbox_inches="tight")
plt.show()
```

A CHART OF THE PRODUCT SOLD 

```python
plt.figure(figsize=(10, 5))

plt.bar(df["Product"], df["Quantity_sold"], color='RED')
plt.xlabel("Product")
plt.ylabel("Quantity Sold")

plt.title("Quantity Sold per Product")
plt.savefig("Sales_charts/Quantity_sold_chart.png", dpi=300, bbox_inches="tight")


plt.show()
```

FIND THE CHEAPEST PRODUCT

```python
cheapest = df.loc[df["Price"].idxmin()]

print("Cheapest Product:", cheapest["Product"])
print("Price: ₦", cheapest["Price"])
```
